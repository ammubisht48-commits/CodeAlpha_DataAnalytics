```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
print("CodeAplha Data Analytics Project Started!")
```

```python
import pandas as pd
Data = {
    "product": ["Laptop","Mobile","Headphones","Laptop","Tablet","Mobile","Headphones","Tablet","Laptop","Mobile"],
    "category": ["Electronics","Electronics","Accessories","Electronics","Electronics","Electronics","Accessories","Electronics","Electronics","Electronics"],
    "Quantity": [2,5,10,1,3,4,6,2,3,5],
    "Price": [50000,20000,1500,50000,25000,20000,1500,25000,50000,20000],
    "Month": ["Jan","Jan","Jan","Feb","Feb","Feb","Mar","Mar","Mar","Mar"]
}
df = pd.DataFrame(Data)
df["Sales"] = df["Quantity"] * df["Price"]
df.to_csv("sales_data.csv",index=False)
print("Dataset created successfully!")
print(df)
```

```python
print("First 5 Rows")
display(df.head())
print("Dataset Shape")
print(df.shape)
print("Column Name:")
print(df.columns)
print("dataset Information")
df.info()
print("Statistical Summary:")
display(df.describe())
```

```python
total_sales = df["Sales"].sum()
print("Total Sales:",total_sales)

total_quantity = df["Quantity"].sum()
print("quantity:",total_quantity)

product_sales = df.groupby("product")["Sales"].sum()
print("\nproduct-wise Sales:")
print("product_sales")

monthly_sales = df.groupby("Month")["Sales"].sum()
print("\nMonthly Sales:")
print("monthly_sales")

best_product = product_sales.idxmax()
print("\nBest product Selling:",best_product)

best_month = monthly_sales.idxmax()
print("\nHighest sales month:",best_month)
```

```python
df.groupby("product")["Sales"].sum().plot(kind="bar")
plt.title("Product-wise sales")
plt.xlabel("product")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()

df.groupby("Month")["Sales"].sum().reindex(["Jan","Feb","Mar"]).plot(kind="line",marker="o")
plt.title("Monthly sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()

df.groupby("category")["Sales"].sum().plot(kind="pie",autopct="%1.1f%%")
plt.title("sales by Category")
plt.ylabel("")
plt.show()
           
```
